```python
# ==========================================
# PROJETO 09: Gestão de Orçamento Familiar
# ==========================================
# Descrição: Sistema simulador de controle financeiro
# com entrada de receitas, despesas e geração de saldo.
# Autor: Desenvolvedor
# Versão: 1.1

# --- 1. CONFIGURAÇÕES E IMPORTAÇÕES ---
import json

# --- 2. MODELAGEM DE DADOS (CLASSE PRINCIPAL) ---
class OrcamentoFamiliar:
    def __init__(self, mes):
        self.mes = mes
        self.receitas = []
        self.despesas = []

    def adicionar_receita(self, descricao, valor):
        self.receitas.append({"descricao": descricao, "valor": valor})

    def adicionar_despesa(self, descricao, valor):
        self.despesas.append({"descricao": descricao, "valor": valor})

    def calcular_resumo(self):
        total_receitas = sum(r["valor"] for r in self.receitas)
        total_despesas = sum(d["valor"] for d in self.despesas)
        saldo_final = total_receitas - total_despesas

        return {
            "mes": self.mes,
            "total_receitas": total_receitas,
            "total_despesas": total_despesas,
            "saldo_final": saldo_final,
            "status_financeiro": "Superávit" if saldo_final >= 0 else "Déficit"
        }

# --- 3. EXECUÇÃO DO FLUXO E TESTE DE SAÍDA ---
if __name__ == "__main__":
    print("=== SIMULADOR DE ORÇAMENTO FAMILIAR - PROJETO 09 ===\n")

    # Criando o orçamento do mês
    orcamento = OrcamentoFamiliar(mes="Outubro")

    # Adicionando receitas de teste
    orcamento.adicionar_receita("Salário Base", 5500.00)
    orcamento.adicionar_receita("Rendimento Extra", 350.00)

    # Adicionando despesas de teste
    orcamento.adicionar_despesa("Aluguel e Condomínio", 1800.00)
    orcamento.adicionar_despesa("Supermercado", 850.00)
    orcamento.adicionar_despesa("Contas de Consumo (Água/Luz/Internet)", 420.00)
    orcamento.adicionar_despesa("Lazer", 300.00)

    # Gerando o resumo de saída
    resumo = orcamento.calcular_resumo()

    # Exibindo a saída formatada de forma legível
    print(f"Resumo Mensal de: {resumo['mes']}")
    print("-" * 45)
    print(f"(+) Total de Receitas: R$ {resumo['total_receitas']:.2f}")
    print(f"(-) Total de Despesas: R$ {resumo['total_despesas']:.2f}")
    print("-" * 45)
    print(f"(=) Saldo Final:      R$ {resumo['saldo_final']:.2f}")
    print(f"Situação Financeira:  {resumo['status_financeiro']}")
    print("=" * 45)
```

```python
# ==========================================
# PROJETO 07: Meu Orçamento Familiar
# ==========================================
# Integrantes: Desenvolvedor
# Descrição: Controle completo de receitas e despesas por categoria

# --- CONSTANTES ---
CATEGORIAS = [
    "Moradia",
    "Alimentação",
    "Transporte",
    "Saúde",
    "Lazer",
    "Outros"
]

# --- BANCO DE DADOS EM MEMÓRIA ---
lancamentos = []

# --- FUNÇÕES DE AUXÍLIO E VALIDAÇÃO ---
def obter_descricao():
    while True:
        desc = input("Digite a descrição: ").strip().capitalize()
        if desc:
            return desc
        print("Erro: A descrição não pode ser vazia.")

def obter_valor():
    while True:
        try:
            valor = float(input("Digite o valor (R$): "))
            if valor > 0:
                return valor
            print("Erro: O valor deve ser maior que zero.")
        except ValueError:
            print("Erro: Digite um número válido.")

def obter_categoria():
    print("\nCategorias Disponíveis:")
    for i in range(len(CATEGORIAS)):
        print(f"{i+1} - {CATEGORIAS[i]}")
    while True:
        try:
            opcao = int(input("Escolha a categoria (1 a 6): "))
            if 1 <= opcao <= 6:
                return CATEGORIAS[opcao - 1]
            print("Erro: Opção inválida. Escolha entre 1 e 6.")
        except ValueError:
            print("Erro: Digite um número inteiro.")

# --- REQUISITOS FUNCIONAIS EM ORDEM (RF01 a RF06) ---

# RF01 - Menu Principal do Sistema
def menu():
    while True:
        print("\n=== MEU ORÇAMENTO FAMILIAR ===")
        print("1 - Registrar receita")
        print("2 - Registrar despesa")
        print("3 - Resumo do mês")
        print("4 - Despesas por categoria")
        print("5 - Listar lançamentos")
        print("0 - Sair")

        try:
            opcao = input("Escolha: ").strip()
            if opcao == "1":
                registrar_receita()
            elif opcao == "2":
                registrar_despesa()
            elif opcao == "3":
                resumo_do_mes()
            elif opcao == "4":
                despesas_por_categoria()
            elif opcao == "5":
                listar_lancamentos()
            elif opcao == "0":
                print("Encerrando o sistema. Até logo!")
                break
            else:
                print("Opção inválida. Escolha uma opção do menu.")
        except Exception as e:
            print(f"Ocorreu um erro inesperado: {e}")

# RF02 - Registrar Receita
def registrar_receita():
    descricao = obter_descricao()
    valor = obter_valor()
    lancamento = {
        "tipo": "receita",
        "descricao": descricao,
        "categoria": "Receita",
        "valor": valor
    }
    lancamentos.append(lancamento)
    print(f"Receita '{descricao}' registrada com sucesso!")

# RF03 - Registrar Despesa
def registrar_despesa():
    descricao = obter_descricao()
    categoria = obter_categoria()
    valor = obter_valor()
    lancamento = {
        "tipo": "despesa",
        "descricao": descricao,
        "categoria": categoria,
        "valor": valor
    }
    lancamentos.append(lancamento)
    print(f"Despesa '{descricao}' registrada na categoria '{categoria}' com sucesso!")

# RF04 - Resumo do Mês
def resumo_do_mes():
    total_receitas = sum(l["valor"] for l in lancamentos if l["tipo"] == "receita")
    total_despesas = sum(l["valor"] for l in lancamentos if l["tipo"] == "despesa")
    saldo = total_receitas - total_despesas

    print("\n--- RESUMO DO MÊS ---")
    print(f"Receitas: R$ {total_receitas:.2f}")
    print(f"Despesas: R$ {total_despesas:.2f}")
    print(f"Saldo: R$ {saldo:.2f}")

    if total_receitas == 0:
        print("Registre ao menos uma receita para analisar o orçamento.")
        return

    comprometimento = (total_despesas / total_receitas) * 100
    print(f"Comprometimento de Renda: {comprometimento:.2f}%")

    if comprometimento <= 70:
        situacao = "Saudável"
    elif comprometimento <= 100:
        situacao = "Atenção – despesas altas"
    else:
        situacao = "Alerta – despesas maiores que a receita"

    print(f"Situação do Orçamento: {situacao}")

# RF05 - Despesas por Categoria
def despesas_por_categoria():
    despesas = [l for l in lancamentos if l["tipo"] == "despesa"]
    if not despesas:
        print("Nenhuma despesa registrada.")
        return

    totais_por_categoria = {cat: 0.0 for cat in CATEGORIAS}
    total_despesas_geral = 0.0

    for d in despesas:
        totais_por_categoria[d["categoria"]] += d["valor"]
        total_despesas_geral += d["valor"]

    print("\n--- DESPESAS POR CATEGORIA ---")
    maior_categoria = ""
    maior_valor = -1.0

    for categoria, total in totais_por_categoria.items():
        if total > 0:
            percentual = (total / total_despesas_geral) * 100
            # Gráfico de barras simples
            grafico = "#" * int(percentual / 5)
            print(f"{categoria:<15}: R$ {total:>8.2f} ({percentual:>5.1f}%) {grafico}")

            if total > maior_valor:
                maior_valor = total
                maior_categoria = categoria

    if maior_categoria:
        print("-" * 45)
        print(f"Categoria de maior gasto: {maior_categoria} (R$ {maior_valor:.2f})")

# RF06 - Listar Lançamentos
def listar_lancamentos():
    if not lancamentos:
        print("Nenhum lançamento registrado.")
        return

    print("\n--- LISTA DE LANÇAMENTOS ---")
    for i in range(len(lancamentos)):
        l = lancamentos[i]
        sinal = "+" if l["tipo"] == "receita" else "-"
        print(f"{i+1}. [{l['categoria']}] {l['descricao']} ({sinal} R$ {l['valor']:.2f})")

# --- EXECUÇÃO DO MENU INTERATIVO ---
if __name__ == "__main__":
    menu()
```
